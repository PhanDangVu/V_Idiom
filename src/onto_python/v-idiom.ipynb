# Khởi tạo Khung và Thành ngữ Mới (Chuẩn hóa Từ Vựng)
Notebook này nạp ontology và áp dụng thuật toán **Tái sử dụng Khung**. Thay vì mỗi cặp thành ngữ tạo ra 1 khung rời rạc, hệ thống sẽ sử dụng Bể tiêu chí chung (Controlled Vocabulary). Những thành ngữ có cùng tổ hợp 4 tiêu chí sẽ tự động được gộp chung vào 1 Khung.

```python
from owlready2 import *
import os

# Đảm bảo nạp đúng file gốc v-idiomV2.rdf (bản sạch nhất)
onto_path = "../../ontology_protege/v-idiomV2.rdf"
if not os.path.exists(onto_path):
    print("Không tìm thấy file ở:", onto_path)
else:
    onto = get_ontology(onto_path).load()
    print("Đã tải thành công ontology:", onto.base_iri)
```

```python
import pandas as pd

# Đọc dữ liệu từ file CSV V3
csv_path = "../../thanh_ngu_tuc_ngu_100_cau_v3.csv"
df = pd.read_csv(csv_path).fillna("")
idioms_data = df.to_dict('records')

print(f"Đã nạp {len(idioms_data)} dòng dữ liệu từ CSV.")
```

```python
# Thuật toán tái sử dụng Khung (Hỗ trợ dữ liệu khuyết)
with onto:
    frame_cache = {}
    frame_counter = 4
    
    for data in idioms_data:
        bc_id = str(data.get("Bối_cảnh", "")).strip()
        hd_id = str(data.get("Hành_động", "")).strip()
        kq_id = str(data.get("Kết_quả", "")).strip()
        md_id = str(data.get("Mục_đích", "")).strip()
        
        signature = (bc_id, hd_id, kq_id, md_id)
        
        if signature in frame_cache:
            khung = frame_cache[signature]
            print(f"[TÁI SỬ DỤNG] '{data['Thành_ngữ_Tiếng_Việt']}' được ghép chung vào {khung.name}")
        else:
            khung = onto.Khung_Bản_Thể_học(f"Khung{frame_counter}")
            
            if bc_id: khung.Có_bối_cảnh = [onto.Bối_cảnh(bc_id)]
            if hd_id: khung.Có_hành_động = [onto.Hành_động(hd_id)]
            if kq_id: khung.Có_kết_quả = [onto.Kết_quả(kq_id)]
            if md_id: khung.Có_mục_đích = [onto.Mục_đích(md_id)]
            
            frame_cache[signature] = khung
            print(f"[TẠO MỚI] '{data['Thành_ngữ_Tiếng_Việt']}' được cấp {khung.name}")
            frame_counter += 1
        
        vn_id = str(data.get("Thành_ngữ_Tiếng_Việt", "")).strip()
        en_id = str(data.get("Thành_ngữ_Tiếng_Anh", "")).strip()
        
        if vn_id:
            vn_idiom = onto.Vietnamese_idiom(vn_id)
            vn_idiom.Có_khung = [khung]
        if en_id:
            en_idiom = onto.English_Idiom(en_id)
            en_idiom.Có_khung = [khung]

print(f"\nĐã xử lý xong {len(idioms_data)} cặp thành ngữ!")
print(f"Tổng số Khung được tạo ra: {len(frame_cache)} khung.")

# Lưu vào file v-idiomV5.rdf
save_path = "../../ontology_protege/v-idiomV5.rdf"
onto.save(file=save_path, format="rdfxml")
print(f"Đã lưu ontology ra file {save_path}")
```
