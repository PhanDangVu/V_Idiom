```python
from owlready2 import *

# 1. Tải ontology
onto = get_ontology("/Users/phandangvu/Desktop/ontology/onto-python/ontology_protege/v-idiomV2.rdf").load()

print("Các class trong ontology:", list(onto.classes()))
```

```python
# 2. Chạy reasoner để thực hiện suy luận (suy luận thuộc tính Có_nghĩa)
with onto:
    # Bạn có thể dùng sync_reasoner() (HermiT) hoặc sync_reasoner_pellet()
    sync_reasoner_pellet(infer_property_values=True, infer_data_property_values=True)

print("\nĐã chạy xong reasoner!")
```

```python
# 3. Tìm và in ra nghĩa của các thành ngữ (dựa trên property chain Có_nghĩa)
for idiom in onto.Idiom.instances():
    print(f"Thành ngữ: {idiom.name}")
    
    # In ra khung ý nghĩa (nghĩa gốc)
    khung = idiom.Có_khung
    if khung:
        print(f" - Khung ý nghĩa: {[k.name for k in khung]}")
        for k in khung:
            if k.Có_bối_cảnh: print(f"   + Bối cảnh: {[x.name for x in k.Có_bối_cảnh]}")
            if k.Có_hành_động: print(f"   + Hành động: {[x.name for x in k.Có_hành_động]}")
            if k.Có_kết_quả: print(f"   + Kết quả: {[x.name for x in k.Có_kết_quả]}")
            if k.Có_mục_đích: print(f"   + Mục đích: {[x.name for x in k.Có_mục_đích]}")
            
    # Các thành ngữ đồng nghĩa (được reasoner suy luận qua property chain)
    dong_nghia = [n.name for n in idiom.Có_nghĩa if n != idiom]
    if dong_nghia:
        print(f" - Đồng nghĩa (suy luận): {dong_nghia}")
    
    print("-" * 50)
```
