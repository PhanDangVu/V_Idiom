```python
# Tạo Individuals dựa trên dữ liệu
with onto:
    for data in idioms_data:
        # 1. Khởi tạo 4 tiêu chí cho khung nghĩa đen
        bc = onto.Bối_cảnh(data["boi_canh"])
        hd = onto.Hành_động(data["hanh_dong"])
        kq = onto.Kết_quả(data["ket_qua"])
        md = onto.Mục_đích(data["muc_dich"])
        
        # 2. Khởi tạo Khung_Bản_Thể_học mới
        khung = onto.Khung_Bản_Thể_học(f"Khung{data['id']}")
        khung.Có_bối_cảnh = [bc]
        khung.Có_hành_động = [hd]
        khung.Có_kết_quả = [kq]
        khung.Có_mục_đích = [md]
        
        # 3. Khởi tạo các thành ngữ và gán khung
        en_idiom = onto.English_Idiom(data["en"])
        en_idiom.Có_khung = [khung]
        
        vn_idiom = onto.Vietnamese_idiom(data["vn"])
        vn_idiom.Có_khung = [khung]

print(f"Đã thêm {len(idioms_data)} cặp thành ngữ và khung thành công!")

# Lưu vào file mới để giữ file gốc
onto.save(file="../../ontology_protege/v-idiomV3.rdf", format="rdfxml")
print("Đã lưu ontology ra file v-idiomV3.rdf")
```
